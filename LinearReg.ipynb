```python
import pandas as pd
import numpy as np
from sklearn import linear_model
```

```python
df = pd.read_csv("/content/hiring.csv")
df
```

```python
import math
median_test_score = math.floor(df['test_score(out of 10)'].mean())
median_test_score
```

```python
import math
mean_test_score = math.floor(df['test_score(out of 10)'].mean())
df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(mean_test_score)
display(df)
```

```python
reg = linear_model.LinearRegression()
```

```python
from word2number import w2n
df['experience'] = df['experience'].apply(lambda x: w2n.word_to_num(x) if isinstance(x, str) else x)
df['experience'] = df['experience'].fillna(0)

reg.fit(df[['experience','test_score(out of 10)','interview_score(out of 10)']],df['salary($)'])
```

```python
df
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
import pandas as pd
new_candidate = pd.DataFrame([[2, 9, 6]], columns=['experience', 'test_score(out of 10)', 'interview_score(out of 10)'])
predicted_salary = reg.predict(new_candidate)
print(f"Predicted salary: ${predicted_salary[0]:.2f}")
```
